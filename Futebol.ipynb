```
import pandas as pd
```

```
tabela = pd.read_csv("/content/drive/MyDrive/DB's/mundo_transfermarkt_competicoes_brasileirao_serie_a.csv.gz")
display(tabela)
```

```
display(tabela.info())
display(tabela.describe())
```

```
qtd_pessoas = tabela["publico"].value_counts()
display(qtd_pessoas)

time = tabela["time_mandante"].value_counts()
display(time)
```

```
import plotly.express as plt

grafico_pub = plt.histogram(tabela, x="time_mandante", y="publico" )
display(grafico_pub)
```

```
grafico_golsFora = plt.histogram(tabela, x="time_visitante", y="gols_visitante" )
display(grafico_golsFora)
```

# Flamengo, São Paulo e Fluminense são muito fortes fora de casa, levando em conta o histórico

```
grafico_golsCasa = plt.histogram(tabela, x = "time_mandante", y = "gols_mandante")
display(grafico_golsCasa)
```

# São Paulo, Flamengo, Atlético-MG, Palmeiras, Grêmio são muito fortes dentro de casa, tendo maior probabilidade de vitória

```
grafico_escanteioCasa = plt.histogram(tabela, x = "time_mandante", y = "escanteios_mandante")
display(grafico_escanteioCasa)
```

# Palmeiras e Flamengo chamam atenção na quantidade de escanteios dentro de casa

```
grafico_escanteioFora = plt.histogram(tabela, x="time_visitante", y = "escanteios_visitante")
display(grafico_escanteioFora)
```

# Atlético Mineiro e Palmeiras tem muitos escanteios fora de casa

```
grafico_faltaFora = plt.histogram(tabela, x="time_visitante", y="faltas_visitante")
display(grafico_faltaFora)
```

# O palmeiras faz muita falta fora de casa

```
grafico_faltaCasa = plt.histogram(tabela, x="time_mandante", y="faltas_mandante")
display(grafico_faltaCasa)
```

# O Internacional, Altético Mineiro, Athlético Paranaense, Fluminense fazem muita falta dentro de casa.

# O que é possível perceber com a análise:
 * Flamengo, São Paulo e Fluminense possuem maior chance de marcar gols fora de casa
 * São Paulo, Flamengo, Atlético-MG, Palmeiras, Grêmio são fortes dentro de casa
 * Palmeiras e Flamengo são fortes em escanteios dentro de casa
 * Atlético Mineiro e Palmeiras tem muitos escanteios fora de casa
 * Palmeiras tem muita falta fora de casa
 * O Internacional, Altético Mineiro, Athlético Paranaense, Fluminense fazem muita falta dentro de casa.
