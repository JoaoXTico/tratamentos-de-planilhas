```python
import pandas as pd
import seaborn as sbn
import matplotlib.pyplot as plt
```

```python
database = pd.read_csv('clientes_banco_analise.csv')
database.info()
```

```python
display(database)
```

```python
display(database.describe())
```

```python
qtde_cancel = database['churn'].value_counts()
display(qtde_cancel)
```

```python
grafico_cancel = qtde_cancel.plot.bar(color=['blue', 'red'])
grafico_cancel.set_title('Quantidade de clientes que cancelaram ou não o serviço')

grafico_cancel.set_xticklabels(['Ativos', 'Cancelaram'], rotation=0)
grafico_cancel.set_xlabel('Churn')
```

```python
idade_cancelados = database[database['churn'] == 1]['idade']
sbn.histplot(idade_cancelados, bins=10, kde=True, color='red')
plt.title('Distribuição de Idade dos Clientes que Cancelaram')
```

```python
sbn.boxplot(data=database, x='renda_mensal', y='escolaridade', palette='Set2')
plt.title('Renda Mensal dos Clientes com Base na Escolaridade')
```
